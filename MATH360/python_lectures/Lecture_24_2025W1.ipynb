```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from scipy.optimize import minimize

print("Packages imported")
```

```python
def gradient_descent(f,grad,x0,alpha=0.1,max_iter=100,eps=1e-4):
    X = np.zeros([max_iter+1,len(x0)])
    X[0,:] = x0
    for k in range(max_iter):
        xk = X[k,:]
        grad_xk = grad(xk)
        if np.linalg.norm(grad_xk) < eps:
            # Truncate the matrix X to include
            # only rows up to current index k
            X = X[:k+1,:]
            break
        X[k+1,:] = xk - alpha*grad_xk
    return X
```

## Gradient Descent for Squared Errors

Let $(\mathbf{x}_0,y_0),\dots,(\mathbf{x}_{N-1},y_{N-1})$ be a set of points such that $\mathbf{x}_k \in \mathbf{R}^n$ for all $k$, and $y_k = 0$ or $y_k = 1$ for all $k$. The sum of squared errors cost function (without regularization) is given by

$$
C(\mathbf{W},b) = \frac{1}{N} \sum_{k=0}^{N-1} (y_k - \sigma(\mathbf{x}_k; \mathbf{W},b) )^2
$$

Compute the partial derivatives $\displaystyle \frac{\partial C}{\partial W_{\ell}}$, $\ell = 0,\dots,N-1$, and $\displaystyle  \frac{\partial C}{\partial b}$.

Compare the partial derivatives above to those for the cross entropy function (without regularization). What do you notice about the terms in the sum for points $(\mathbf{x}_k,y_k)$ such that $\sigma(\mathbf{x}_k; \mathbf{W},b) \approx 1$ and $y_k = 0$. That is, when the model $\sigma(\mathbf{x}; \mathbf{W},b)$ is misclassfying the target $y_k$. What about when $\sigma(\mathbf{x}_k; \mathbf{W},b) \approx 0$ and $y_k = 1$. How does that effect gradient descent?

Hint: If the parameters $\mathbf{W}$ and $b$ are near optimal then $C(\mathbf{W},b)$ is near a minimum value and $\nabla C \approx 0$. But if the parameters $\mathbf{W}$ and $b$ are *not* near optimal then want to take a big step in the direction $\nabla C$ towards a minimum.

Note: The (bad) properties of the gradient of the sum of sqaured errors and the (good) properties of cross-entropy are the reason we use cross-entropy instead of squared errors to fit logistic regression models.
